# 4. Padronização das Variáveis

### 📌 Descrição das Variáveis

| Variável | Descrição | O que esperar na análise |
|---|---|---|
| **inadimplente_2_anos** | Variável alvo. Indica se o cliente apresentou inadimplência grave, com atraso de 90 dias ou mais, dentro de 2 anos. | Variável binária (`0` = não inadimplente, `1` = inadimplente). É importante verificar o **balanceamento das classes**. |
| **utilizacao_credito** | Proporção do limite de crédito rotativo utilizado pelo cliente, principalmente cartões e linhas de crédito sem garantia. | Valores maiores podem indicar maior pressão financeira. Espera-se concentração em valores baixos, mas podem existir **valores acima de 1 e outliers**. |
| **idade** | Idade do cliente em anos. | Deve apresentar valores inteiros positivos. Vale verificar **idades impossíveis ou muito extremas** e analisar a relação entre faixa etária e inadimplência. |
| **atrasos_30_59_dias** | Quantidade de vezes que o cliente apresentou atraso entre 30 e 59 dias. | Espera-se grande concentração em `0`. Valores elevados podem indicar maior risco e também devem ser investigados como possíveis **outliers/anomalias**. |
| **razao_divida** | Relação entre as obrigações financeiras do cliente e sua capacidade financeira/renda. | Distribuição provavelmente assimétrica, com possibilidade de **valores extremamente altos**. Pode ser um indicador importante de endividamento. |
| **renda_mensal** | Renda mensal declarada do cliente. | Deve possuir forte assimetria, valores muito altos e **dados ausentes**. Pode ser necessário tratamento de missing values e outliers. |
| **linhas_credito_abertas** | Número de empréstimos e linhas de crédito atualmente abertas. | Normalmente valores inteiros e positivos. Pode ajudar a identificar clientes com elevada exposição ao crédito. |
| **atrasos_90_dias** | Número de vezes em que o cliente apresentou atraso de 90 dias ou mais. | Deve ter forte concentração em `0`. Provavelmente será uma das variáveis mais relacionadas ao risco de inadimplência. |
| **emprestimos_imobiliarios** | Número de empréstimos ou linhas de crédito relacionados a imóveis. | Espera-se concentração em poucos empréstimos (`0`, `1`, `2` etc.). Valores muito elevados devem ser investigados. |
| **atrasos_60_89_dias** | Quantidade de atrasos entre 60 e 89 dias. | Assim como as demais variáveis de atraso, espera-se predominância de `0` e relação positiva com risco de inadimplência. |
| **numero_dependentes** | Número de dependentes financeiros do cliente. | Espera-se valores inteiros pequenos. Existem **dados ausentes**, que precisarão ser analisados durante a preparação. |

```python
dados.columns = ["inadimplente_2_anos", "utilizacao_credito", "idade", "atrasos_30_59_dias", "razao_divida",
                 "renda_mensal", "linhas_credito_abertas", "atrasos_90_dias", "emprestimos_imobiliarios",
                 "atrasos_60_89_dias", "numero_dependentes"]
dados.columns.tolist()
```

# 7. Análise dos Valores Aunsentes

## 7.1 - Renda Mensal

```python
dados['renda_mensal'].describe().round(2)
```

```python
df_renda = dados.copy()
df_renda["renda_mensal_ausentes"] = df_renda['renda_mensal'].isna()
```

```python
df_renda.groupby("renda_mensal_ausentes")[
    [
        "utilizacao_credito", "idade", "razao_divida", "linhas_credito_abertas",
        "atrasos_30_59_dias", "atrasos_90_dias", "emprestimos_imobiliarios"
    ]
].agg(["mean", "median"])
```

```python
df_renda.groupby("renda_mensal_ausentes")["inadimplente_2_anos"].agg(
    quantidade="count",
    inadimplentes="sum",
    taxa_inadimplencia="mean"
)
```

```python
dados.groupby(
    dados["renda_mensal"].isna()
)["razao_divida"].describe().round(2)
```

```python
dados.groupby(
    dados["renda_mensal"].isna()
)["razao_divida"].apply(lambda x: (x > 1).mean() * 100).round(2)
```

```python
dados.groupby(
    [dados["renda_mensal"].isna(), "inadimplente_2_anos"]
)["razao_divida"].median().round(2)
```

### Conclusão — Renda Mensal

A variável `renda_mensal` apresenta aproximadamente **19,82% de valores ausentes** e uma distribuição assimétrica, com presença de valores extremos.

A análise mostrou que os clientes com renda ausente apresentam características diferentes dos demais, principalmente em relação à `razao_divida`, além de uma taxa de inadimplência de aproximadamente **5,61%**, contra **6,95%** entre os clientes com renda informada.

Também foi identificado que a `razao_divida` apresenta comportamento bastante distinto quando a renda está ausente, indicando que valores elevados nessa variável não devem ser tratados automaticamente como outliers.

Dessa forma, os registros com renda ausente serão mantidos e tratados posteriormente durante o pré-processamento, utilizando imputação pela **mediana**, calculada apenas sobre os dados de treinamento.

## 7.2 - Número Dependentes

```python
dados['numero_dependentes'].describe().round(3)
```

```python
df_dependentes = dados.copy()
df_dependentes["numero_dependentes_ausentes"] = df_dependentes["numero_dependentes"].isna()
```

```python
df_dependentes.groupby("numero_dependentes_ausentes")[
    [
        "utilizacao_credito", "idade","razao_divida", "linhas_credito_abertas",
        "atrasos_30_59_dias", "atrasos_90_dias", "emprestimos_imobiliarios"
    ]
].agg(["mean", "median"])
```

```python
df_dependentes.groupby("numero_dependentes_ausentes")["inadimplente_2_anos"].agg(
    quantidade="count",
    inadimplentes="sum",
    taxa_inadimplencia="mean"
)
```

```python
dados.groupby(
    dados["numero_dependentes"].isna()
)["razao_divida"].describe().round(2)
```

```python
pd.crosstab(
    dados["numero_dependentes"].isna(),
    dados["renda_mensal"].isna(),
    normalize="index"
).mul(100).round(2)
```

### Conclusão — Número de Dependentes

A variável `numero_dependentes` possui aproximadamente **2,62% de valores
ausentes**.

A análise mostrou que clientes sem essa informação apresentam perfil
diferente dos demais, incluindo menor taxa de inadimplência.

Além disso, foi identificado que **100% dos registros com número de
dependentes ausente também possuem renda mensal ausente**, indicando forte
sobreposição entre os padrões de ausência dessas duas variáveis.

Dessa forma, os registros serão mantidos e os valores ausentes serão
tratados posteriormente no pré-processamento, utilizando imputação pela
mediana e preservando, se necessário, a informação de ausência por meio
de uma variável indicadora.

## 8. Tratamento de Valores Inválidos — Idade

```python
dados["idade"].describe()
```

```python
idade_18 = dados[dados['idade']<18]
len(idade_18)
```

```python
idade_18
```

```python
dados = dados[dados['idade']>=18]
```

### Conclusão — Idade

Foi identificado **1 registro com idade igual a 0**, valor considerado inválido para o contexto da análise de risco de crédito.

Como o registro representa uma parcela insignificante da base e não é possível determinar a idade correta do cliente, optou-se pela sua remoção.

Após o tratamento, foram mantidos apenas clientes com idade igual ou superior a **18 anos**.
